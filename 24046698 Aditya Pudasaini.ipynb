# 2. Data Preperation

## Importing necessary libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import ttest_ind as ttest
import seaborn as sns 
```

## Loading data into pandas dataframe and displaying last 50 records

```python
df = pd.read_csv("ids.csv", index_col=0) # Reads the file and loads the data as a daaframe into df
df.tail(50) # Displays the last 50 records
```

## Creating a separate Data Frame containing only the specific columns

```python
columns = ['Destination Port', 'Flow Duration', 'Total Fwd Packets', 'Total Backward Packets', 
           'Total Length of Fwd Packets', 'Total Length of Bwd Packets', 'Fwd Packet Length Mean', 
           'Bwd Packet Length Mean', 'Flow Bytes/s', 'Flow Packets/s', 'Packet Length Mean', 
           'Packet Length Std', 'Average Packet Size', 'Active Mean', 'Idle Mean', 'Label']
modified_df = df[columns] # Creating a new dataframe that stores the values of specific coulumns
modified_df
```

## Checking for NaN (missing values) and remove them from the newly created DataFrame.

```python
print(modified_df.isnull().sum())

modified_df = modified_df.dropna()

print("\n",modified_df.isnull().sum()) 
```

## Checking if duplicated records exist and finding their total number

```python
duplicates = modified_df.duplicated().sum() # Finds the number of duplicate records and adding uo all of them
print("Total number of duplicates records are: ", duplicates)
```

## Showing total number of columns and displaying name of all columns

```python
num_of_columns = modified_df.shape[1] # Shows the number of columns since shape[1] is used as only using shape will give the number of rows and columns
print("Number of columns: ", num_of_columns)
print(modified_df.columns) # Printing the name of columns in the modified dataframe
```

## Renaming the values in the Label column as Normal traffic for BENIGN and Infiltration as Attack

```python
modified_df['Label'] = modified_df['Label'].replace({'BENIGN':'Normal Traffic','Infiltration':'Attack'}) # Replacing the values in the Label column
modified_df
```

# 3. Data Analysis 

## Showing summary of various statistics for two columns

```python
flowDuration = modified_df['Flow Duration']
print(flowDuration.agg(["sum", "mean", "std", "skew", "kurt"])) # SHowing statistics of a Flow Duration column
```

```python
averagePacketSize = modified_df['Average Packet Size']
print(averagePacketSize.agg(["sum", "mean", "std", "skew", "kurt"])) # SHowing statistics of a Average Packet Size column
```

## Calculating correlation

```python
correlationTable = modified_df.corr(numeric_only=True)

correlation = correlationTable.unstack()
# unstack turns the data which were arranged in matrix from when we used corr, to a pair like A B, A C

correlation = correlation[(correlation < 0.999) & (correlation != 1)]
# Making sure there are no perfect correlation available

correlation = correlation.drop_duplicates()
# Removing any duplicates form occuring

print(correlation.sort_values(ascending=False).head(5))
# Printing only the top 5 by combing sort_values to sort and head to display
```

```python
plt.figure(figsize=(12, 10))
sns.heatmap(correlationTable, annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.tight_layout()
```

## Checking whether the Label, as the dependent variable, is balanced or not

```python
makingBarChart = modified_df['Label'].value_counts() # Calculates how many times each value in label appears

makingBarChart.plot(kind='bar') # Plotting bar chart 

plt.title("Label Distribution") # Naming the title
plt.xlabel("Category") 
plt.ylabel("Frequency")
```

## Determining which category in the Label column has the highest Average Flow Duration and the highest average Packet Length Mean.

```python
makingPieChart = modified_df.groupby('Label').mean(numeric_only=True)

makingPieChart['Flow Duration'].plot(kind='pie', autopct='%1.2f%%')
plt.title("Average Flow Duration by Label")
```

```python
makingPieChartAgain = modified_df.groupby('Label').mean(numeric_only=True)

makingPieChartAgain['Packet Length Mean'].plot(kind='pie',autopct='%1.2f%%')
plt.title("Average Packet Length Mean by Label")
```

```python
figure, axis = plt.subplots(figsize=(8, 5))
# Creates a blank canvas where the figure will be drawn with 8 and 5 units wide and height

modified_df.boxplot(column='Fwd Packet Length Mean', by='Label', ax = axis)
# Generate a boxplot of Fwd Packet Length Mean grouped by Label

#Naming the title and label
axis.set_title("Fwd Packet Length Mean by Label")
axis.set_xlabel("Label")
axis.set_ylabel("Fwd Packet Length Mean")
```

```python
figure, axis = plt.subplots(figsize=(8, 5))
sns.boxplot(x='Label', y='Fwd Packet Length Mean', data=modified_df, ax=axis, showfliers=False)
axis.set_title("Fwd Packet Length Mean by Label")
axis.set_xlabel("Label")
axis.set_ylabel("Fwd Packet Length Mean")
plt.show()
```

```python
normalTraffic = modified_df[modified_df['Label'] == 'Normal Traffic']['Flow Duration']
attack = modified_df[modified_df['Label'] == 'Attack' ]['Flow Duration']
# Stores the values of rows in which the Label is NormaL Traffic under Flow Duration column

t_stat, p_value = ttest(normalTraffic, attack, equal_var=False)
# We run a t-test in which we compare the average flow duration for normal tarffic and attack
# We store those values in t_stat which stores the difference between their mean
# p_value tells whether the difference is statistically significant or due to chance

print("P-value:", p_value)

# Checking the difference
if p_value < 0.05:
    print("Significant difference")
else:
    print("No significant difference")
```
